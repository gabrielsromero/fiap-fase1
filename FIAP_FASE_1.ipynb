## 1.3. CÓDIGO EM PYTHON
Este script implementa em Python a lógica definida na Etapa 2. Ele percorre cada leitura da telemetria, compara os valores com as faixas seguras e imprime a decisão de cada uma: LANÇAR (todos os parâmetros normais) ou ABORTAR (qualquer parâmetro fora da faixa).

````python
contador = 0
for leitura in leituras:
  contador = contador + 1
  if (leitura["temp_interna"] >= temp_interna_min and leitura["temp_interna"]  <= temp_interna_max
    and leitura["temp_externa"]  >= temp_externa_min and leitura["temp_externa"] <= temp_externa_max
    and leitura["integridade_estrutural"] == integridade_estrutural_pad
    and leitura["energia"]  >=  energia_pad
    and leitura["pressao_tanques"]  >=  pressao_tanques_min and leitura["pressao_tanques"] <= pressao_tanques_max
    and leitura["modulos_criticos"] ==  modulos_criticos_pad):
    resultado = "LANÇAR"
  else:
    resultado = "ABORTAR"

  print(f" Leitura {contador} : {resultado}")
````

## 1.4. Análise Energética

Nesta etapa calculamos a **autonomia energética inicial** da nave: com base na energia realmente disponível a bordo, estimamos quantos ciclos de decolagem a espaçonave consegue sustentar antes de precisar recarregar.

O cálculo parte de quatro fatores: a capacidade total das baterias, a carga atual (obtida da telemetria), as perdas energéticas do sistema e o consumo estimado em cada decolagem. A partir deles, chegamos ao número de ciclos de autonomia.

### Premissas adotadas

Por se tratar de uma missão simulada, os valores abaixo são premissas justificadas do projeto:

| Fator | Valor | Justificativa |
|---|---|---|
| Capacidade total das baterias | 500 kWh | Banco de baterias de uma nave de exploração de porte médio. |
| Carga atual | 98% | Estado inicial da missão, obtido da 1ª leitura da telemetria. |
| Perdas energéticas | 10% | Perdas típicas por calor, atrito e ineficiência de conversão. |
| Consumo por decolagem | 100 kWh | A decolagem é o evento de maior gasto energético da operação. |

### Método de cálculo

1. **Energia disponível** = capacidade total × carga atual (%)
2. **Energia útil** = energia disponível − perdas (%)
3. **Autonomia** = energia útil ÷ consumo por decolagem

````python
capacidade_total = 500
carga_atual = leituras[0]["energia"] /100
perdas = 10 / 100
consumo_decolagem = 100
````

````python
energia_disponivel = capacidade_total * carga_atual
energia_util = energia_disponivel - (energia_disponivel * perdas)
autonomia = energia_util // consumo_decolagem

print(f"Autonomia inicial: {autonomia} ciclos completos de decolagem")
print(f"A nave pode realizar {autonomia} decolagens seguras antes de exigir recarga.")
````
